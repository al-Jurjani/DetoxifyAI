```python
# --- Step 1: Import libraries
import pandas as pd
```

```python
# --- Step 2: Load the datasets
# Replace with your file paths if needed
text_df = pd.read_csv("../Data/kaggle_dataset_test.csv")
labels_df = pd.read_csv("../Data/kaggle_dataset_test_labels.csv")
```

```python
# --- Step 3: Inspect the structure
print("Text data:")
print(text_df.head())
print("\nLabel data:")
print(labels_df.head())
```

```python
# --- Step 4: Merge datasets on 'id'
merged_df = pd.merge(text_df, labels_df, on="id")
```

```python
# --- Step 5: Combine all label columns into a single binary label
# Typically, Kaggle toxic dataset labels include:
# toxic, severe_toxic, obscene, threat, insult, identity_hate
label_cols = ['toxic', 'severe_toxic', 'obscene', 'threat', 'insult', 'identity_hate']
```

```python
merged_df.head(25)
```

```python
merged_df.shape # type: ignore
```

```python
# --- Step 5: Remove unlabeled rows (-1 means not used for scoring)
merged_df = merged_df[~(merged_df[label_cols] == -1).any(axis=1)]
```

```python
merged_df.shape # type: ignore
```

```python
merged_df.head(15)
```

```python
# # Replace -1 with NaN if applicable (in test_labels)
# merged_df[label_cols] = merged_df[label_cols].replace(-1, 1)
```

```python
# Create the single binary label column
merged_df['label'] = (merged_df[label_cols].sum(axis=1) > 0).astype(int)
```

```python
# --- Step 6: Keep only necessary columns
final_df = merged_df[['comment_text', 'label']]
```

```python
final_df.rename(columns={'comment_text': 'text'})
```
